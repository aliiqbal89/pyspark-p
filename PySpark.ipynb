Introduction to PySpark:


What Is PySpark?


PySpark is the python API for Apache Spark, an open source use for big data processings and analytics.

Apache spark is written in scala and java
pyspark allow you to write spark application using python, making easy for python developer to handle large datasets.

Why Pyspark?

scalable: can handle terabyte or petabyte of data

Fast: Uses in memory computing for better speed than hadoop

Python-Friendly: use python library like pandas,numpyalong with pyspark

Distributed processing: Break the data into chunks and processs in parallel across a cluster


Pyspark vs Pandas

Processing: PySpark distributed pandas single machine

speed pyspark faster on big data pandas slower on big data

data size Pyspark handles large dataset pandas limited to RAM

Real World use case:

Recommendation system (netflix , Amazon )

Real time analytics for websites and apps

big data ETL pipeline

SparkSession:
Before working with Pyspark, you must create a SparkSession. This Is your entry point to spark functionally

```python
from pyspark.sql import SparkSession
spark=SparkSession.builder.appName("SimpleApp").getOrCreate()
```

```python
#create data

data=[("Ali",23),("Ahmed",30),("Mohd",28)]
columns=["Name","age"]

# dataframe

df=spark.createDataFrame(data,columns)

# show data

df.show()

```

```python
#create data

data=[
    ("Ali",25,"New York"),
    ("Mohd",30,"India"),
    ("Ahmed",49,"pak"),
    ("Abdullah",12,"uk"),
    ("Sam",19,"china"),

]

columns=["Name","Age","City"]

df=spark.createDataFrame(data,schema=columns)
df.show()




```

Show schema

```python
df.printSchema()
```

select columns

```python
df.select("Name","City").show()
```

filter rows

```python
df.filter(df.Age>21).show()
```

add new column

```python
from pyspark.sql.functions import col
df=df.withColumn("AgePlus10",col("Age")+10)
df.show()
```

Rename column

```python
df=df.withColumnRenamed("City","Location")
df.show()
```

Drop column

```python
df = df.drop("AgePlus10")
df.show()
```

Sort data

```python
df.orderBy("Age").show()
df.orderBy(df.Age.desc()).show()
```

Group and aggregation

```python
df.groupBy("Location").count().show()
df.groupBy("Location").avg("Age").show()
```

```python
df.describe().show()
```

show people whose name start with 'A' or 'D'

```python
df.filter(df.Name.startswith("A")| df.Name.startswith("D")).show()
```

Create a new column "Seinor" (if age>30)

```python
df.withColumn("Senior",col("Age")>30).show()
```

find the youngest and oldest person

```python
from pyspark.sql.functions import min, max
df.select(min("Age"),max("Age")).show()
```

Capitalize all city names

```python
from pyspark.sql.functions import upper
df.withColumn("CityCapital",upper(df.Location)).show()
```

show only those people whose name have exactly 4 letters

```python
from pyspark.sql.functions import length
df.filter(length("Name")==4).show()
```

Create a column "Age Group" with the following

if age < 25 --> "Young"

if age 25 <= age < 30 --> "Adult"

if age>=30 --> "Senior"

```python
from pyspark.sql.functions import when

df.withColumn("AgeGroup",
    when(df.Age < 25, "Young")
    .when((df.Age >= 25) & (df.Age < 30), "Adult")
    .otherwise("Senior")
).show()
```

count total people from new york

```python
df.filter(df.Location=="New York").count()
```

replace "china" with Chinese

```python
from pyspark.sql.functions import when

df.withColumn("CityUpdated",
    when(df.Location == "china", "Chinese")
    .otherwise(df.Location)
).show()
```

find the name of the oldest person

```python
df.orderBy(df.Age.desc()).select("Name").first()[0]
```

Show people living in cities that contain the word "New"

```python
df.filter(df.Location.contains("New")).show()
```

display all name in alphabetical order

```python
df.orderBy("Name").select("Name").show()
```

count the number of unique cities

```python
df.select("Location").distinct().count()
```

count number of letter in each person name

```python
df.withColumn("NameLength",length("Name")).show()
```

Show the averages name length of all people

```python
from pyspark.sql.functions import length

df.withColumn("NameLength", length("Name")) \
  .selectExpr("avg(NameLength) as AvgNameLength") \
  .show()
```

Show people whose name end with e

```python
df.filter(df.Name.endswith("d")).show()
```

Reverse each person name.

```python
from pyspark.sql.functions import reverse

df.withColumn("ReverseName",reverse("Name")).show()
```

check if all people are older than 18

```python
df.filter(df.Age<=18).count()==1
```

```python
min_age=df.agg(min("Age")).collect()[0][0]
all_above = min_age >30
print("all",all_above)
```

extract the last 2 letter of each city

```python
from pyspark.sql.functions import substring
df2 = df.withColumn("last_2_letters", substring("Location", -2, 2))
df2.show()
```

show cities in uppercase and lowercase

```python
from pyspark.sql.functions import upper, lower
df_with_cases = df.withColumn("city_upper", upper("Location")) \
                  .withColumn("city_lower", lower("Location"))
df_with_cases.show()
```

show if any person is exactly 30 year old

```python
df_30 = df.filter(col("age") == 30)
df_30.show()
```

Find the name of the 2nd oldest person

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import dense_rank

window = Window.orderBy(df["Age"].desc())

df.withColumn("Rank", dense_rank().over(window)) \
  .filter("Rank = 2") \
  .select("Name", "Age") \
  .show()
```
